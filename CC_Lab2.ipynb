```python
roll_no = "1024160122"
digits = [int(d) for d in roll_no]
L = [d * 10 for d in digits]
print(L)
L.append(500)
# append() adds an element at the end of the list.
print(L)
L.insert(3, 250)
print(L)
# insert() adds an element at a specified index, need to provide position.
L.remove(20)
print(L)
removed = L.pop(4)
print(L)
print(removed)
L.sort()
print(L)
L.sort(reverse=True)
print(L)
print(L[:3])
print(L[-3:])
average = sum(L) / len(L)
greater_than_avg = [x for x in L if x > average]
print(average)
print(greater_than_avg)
```

```python
scores = tuple(L[:8])
print(scores)
highest = max(scores)
lowest = min(scores)
print(highest)
print(scores.index(highest))
print(lowest)
print(scores.count(lowest))
# Tuples cannot be reversed in place because they are immutable.
reversed_list = list(reversed(scores))
print(reversed_list)
value = int(input("\nEnter a score to search: "))
if value in scores:
    print(scores.index(value))
else:
    print("Score not present.")
try:
    scores[0] = 100
except TypeError as e:
    print("\nError:", e)
# Tuples are immutable, while lists are mutable.
first, second, *remaining = scores
print("\nFirst:", first)
print("Second:", second)
print("Remaining:", remaining)
```

```python
import random
random.seed(1024160122)
numbers = [random.randint(100, 900) for _ in range(100)]
print(numbers)
odd = [x for x in numbers if x % 2 != 0]
print(len(odd))
even = [x for x in numbers if x % 2 == 0]
print(len(even))
def isPrime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
prime_list = [x for x in numbers if isPrime(x)]
print(len(prime_list))
print(prime_list)
frequency = {}
for num in numbers:
    frequency[num] = frequency.get(num, 0) + 1
most = max(frequency, key=frequency.get)
print(most)
print(frequency[most])
```

```python
digits = [int(d) for d in "1024160122"]
A = {d * 7 for d in digits}
B = {d * 9 for d in digits}
print(A)
print(B)
print(A.union(B))
print(A.intersection(B))
print(A.difference(B))
print(B.difference(A))
# difference() returns elements present in one set only,
# while symmetric_difference() returns elements present in either set but not both.
print(A.symmetric_difference(B))
print(A.issubset(B))
print(A.issuperset(B))
x = int(input("\nEnter value to remove from A: "))
A.discard(x)
print(A)
```

```python
my_dict = {
    "name" : "Riya Gupta",
    "roll_no" : "1024160122",
    "branch" : "CSE",
    "age" : 21,
    "city" : "Ludhiana"
}
print(my_dict)
my_dict["location"] = my_dict.pop("city")
print(my_dict)
my_dict["cgpa"] = 9.66
print(my_dict)
my_dict["age"] += 1
print(my_dict)

dict1 = my_dict.copy()
dict2 = my_dict.copy()
removed = dict1.pop("branch")
print(dict1)
print(removed)
del dict2["branch"]
print(dict2)
# pop() removes and returns the value,
# whereas del only removes the key-value pair.
for key, value in my_dict.items():
    print(key, "->", value)
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Email not found.")

friend_dict = {
    "name": "xyz",
    "roll_no": "1024160029",
    "branch": "COSE",
    "age": 20,
    "city": "Patiala"
}
merged = {**my_dict, **friend_dict}
print(merged)
# when duplicates exists then second dict overwrites the first one
string_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print(string_dict)
```
