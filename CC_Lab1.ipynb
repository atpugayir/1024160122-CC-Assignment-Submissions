```python
def is_prime(num):
  if num < 2:
    return False
  for i in range(2, int(num**0.5) + 1):
    if num % i == 0:
      return False
  return True

n = int(input("Enter a number (n) to sum prime numbers up to n: "))
sum = 0
for i in range(1, n + 1):
  if is_prime(i):
    sum += i
print(f"The sum of prime numbers from 1 to {n} is: {sum}")
```

```python
def sum_odd_numbers(n):
  total_sum = 0
  for i in range(1, n + 1):
    if i % 2 != 0:
      total_sum += i
  return total_sum

n = int(input("Enter a number (n) to sum all odd numbers up to n: "))
result = sum_odd_numbers(n)
print(f"The sum of odd numbers from 1 to {n} is: {result}")
```

```python
def is_prime(num):
  if num < 2:
    return False
  for i in range(2, int(num**0.5) + 1):
    if num % i == 0:
      return False
  return True

def sum_prime_numbers(n):
  total_sum = 0
  for i in range(1, n + 1):
    if is_prime(i):
      total_sum += i
  return total_sum

n_prime_func = int(input("Enter a number (n) to sum all prime numbers up to n: "))
result_prime_func = sum_prime_numbers(n_prime_func)
print(f"The sum of prime numbers from 1 to {n_prime_func} is: {result_prime_func}")
```
